## Loading model
Here we will load a pretrained word2vec model from google

google full pre-trained model available at https://drive.google.com/file/d/0B7XkCwpI5KDYNlNUTTlSS21pQmM/edit (1.5 GB)

a smaller model (345 MB unpacked), which is the one we'll be using, is available at https://github.com/eyaler/word2vec-slim/blob/master/GoogleNews-vectors-negative300-SLIM.bin.gz

```python
import gensim
model = gensim.models.KeyedVectors.load_word2vec_format('GoogleNews-vectors-negative300-SLIM.bin.gz', binary=True)  
```

```python
import pandas as pd
```

```python
data = pd.read_csv("product_info.csv",index_col=0)
```

```python
data.head(10)
```

```python
data['product_name'].shape
```

```python
# preprocess using gensim.utils.simple_preprocess function
data['product_name_preprocessed'] = data['product_name'].apply(lambda x: gensim.utils.simple_preprocess(x))
```

```python
data['product_name_preprocessed'].head(10)
```

```python
# build vocabulary and train model
model = gensim.models.Word2Vec(
    data['product_name_preprocessed'].tolist(),
    window=10,
    min_count=3,
    workers=-1,
   )
```

```python
# Done building the model, let's look at some results
w1 = "dior"
model.wv.most_similar(positive=w1)
```

```python
# look up top 6 words similar to 'science'
w2 = ["perfume"]
model.wv.most_similar(positive=w2,topn=6)
```

```python
# look up top 6 words similar to 'france'
w3 = ["serum"]
model.wv.most_similar(positive=w3,topn=6)
```
