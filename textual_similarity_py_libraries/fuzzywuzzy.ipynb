```python
from fuzzywuzzy import fuzz
from fuzzywuzzy import process
import pandas as pd
```

```python
hospital_accounts = pd.read_csv('hospital_account_info.csv')
hospital_reinbursement = pd.read_csv('hospital_reimbursement.csv')
```

```python
results = []

for index, row in hospital_accounts.iterrows():
    match = process.extractOne(row['Facility Name'], hospital_reinbursement['Provider Name'])
    if match[1]>=80:
        results.append((row['Facility Name'], match[0], match[1]))

matched_df = pd.DataFrame(results, columns=['Facility Name', 'Provider Name', 'similarity_score'])
```

```python
matched_df.sort_values(by=['similarity_score'], ascending=False)
```
