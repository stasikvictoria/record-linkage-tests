```python
hospital_accounts = pd.read_csv('hospital_account_info.csv')
hospital_reinbursement = pd.read_csv('hospital_reimbursement.csv')
```

```python
# defining which column match with which column (because they do not have the same name)
left_on = ["Facility Name", "Address", "City", "State"]
right_on = ["Provider Name", "Provider Street Address", "Provider City", "Provider State"]
```

```python
# we use the fuzzymatcher library to find matches

matched_results = fuzzymatcher.fuzzy_left_join(hospital_accounts,
                                               hospital_reinbursement,
                                               left_on,
                                               right_on, left_id_col='Account_Num',
                                               right_id_col='Provider_Num')
```

```python
# we reorganize the columns and we sort the result
cols = ["best_match_score", "Facility Name", "Provider Name", "Address", "Provider Street Address", "Provider City", "City", "Provider State", "State"]
matched_results[cols].sort_values(by=['best_match_score'], ascending=False).head(5)
```

```python
# observing the worst matching scores
matched_results[cols].sort_values(by=['best_match_score'], ascending=True).head(5)
```

One of the datasets contains data from Puerto Rico and the other one does not. It has create mismatches. This is why it is important to understand our data before performing the matching process. Sometimes, some cleaning is needed.

```python
# looking at matches with a score that are not too high and not too low
matched_results[cols].query("best_match_score <= 0.80").sort_values(by=['best_match_score'], ascending=False).head(5)
```

We can see that the matches are more ambiguous : are ADVENTIST HEALTH UKIAH VALLEY and UKIAH VALLEY MEDICAL CENTER the same thing? Once again you need to really understand the data to answer this question.
