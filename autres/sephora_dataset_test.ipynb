# Gensim

```python
df["product_name"].shape
```

```python
df['product_name_processed'] = df['product_name'].apply(lambda x: gensim.utils.simple_preprocess(x))
```

```python
df['product_name_processed'].head(20)
```

```python
model = gensim.models.Word2Vec(
    df['product_name_processed'].tolist(),
    window=10,
    min_count=1,
    workers=-1
)
```

```python
w1 = "porefessional"
model.wv.most_similar(positive=w1)
```

```python
query = "porefessional"
query_processed = gensim.utils.simple_preprocess(query)
```

```python
query_embedding = model.wv[query_processed]
```

```python
df['similarity'] = df['product_name_processed'].apply(
    lambda x: model.wv.n_similarity(query_processed, x)
)
```

```python
most_similar_documents = df.sort_values(by='similarity', ascending=False)['product_name'].tolist()
```

```python
print(most_similar_documents)
```

```python
# Step 1: Preprocess the query
query = "miss dior"
query_processed = gensim.utils.simple_preprocess(query)

# Step 2: Get the Word2Vec embedding for the preprocessed query
query_embedding = model.wv[query_processed]

# Step 3: Calculate similarity between the query embedding and product name embeddings
df['similarity'] = df['product_name_processed'].apply(
    lambda x: model.wv.n_similarity(query_processed, x)
)

# Step 4: Sort the DataFrame based on similarity and get the most similar documents
result_df = df.sort_values(by='similarity', ascending=False)

# Extract the most similar documents and their similarity scores
most_similar_documents = result_df['product_name'].tolist()
similarity_scores = result_df['similarity'].tolist()

# Print or use the most similar documents and their similarity scores
for doc, score in zip(most_similar_documents, similarity_scores):
    if score >= 0.5:
        print(f"Document: {doc}, Similarity Score: {score}")

```
