# Introduction

```python
from urllib.request import urlopen
html = urlopen("http://python.org/").read()
print(len(html))
```

```python
tokens = [tok for tok in html.split()]
print ("Total no of tokens : " + str(len(tokens)))
```

```python
print(tokens[0:100])
```

```python
import re
html_string = html.decode('utf-8')
tokens = re.split('\W+', html_string)
print(len(tokens))
```

```python
print(tokens[0:100])
```

```python
from bs4 import BeautifulSoup
import nltk
```

```python
soup = BeautifulSoup(html, 'html.parser')
clean = soup.get_text()
tokens = [tok for tok in clean.split()]
print(tokens[:100])
```

```python
import operator
freq_dis={}
for tok in tokens:
    if tok in freq_dis:
        freq_dis[tok] += 1
    else:
        freq_dis[tok] = 1
sorted_freq_dist = sorted(freq_dis.items(), key=operator.itemgetter(1), reverse=True)
print(sorted_freq_dist[:25])
```

```python
Freq_dist_nltk = nltk.FreqDist(tokens)
print(Freq_dist_nltk)
```

```python
for k,v in Freq_dist_nltk.items():
    print(str(k) +':'+ str(v))
```

```python
Freq_dist_nltk.plot(50, cumulative=False)
```

```python
inputstring = ' This is an example sent. The sentence splitter will split on sent markers. Ohh really!!!'
from nltk.tokenize import sent_tokenize
all_sent = sent_tokenize(inputstring)
print(all_sent)
```

```python
s = "Hi Everyone ! hola gr8"
print(s.split())
```

```python
from nltk.tokenize import regexp_tokenize, wordpunct_tokenize, blankline_tokenize
print(regexp_tokenize(s, pattern='\w+'))
```

```python
print(regexp_tokenize(s, pattern='\d+'))
```

```python
print(wordpunct_tokenize(s))
```

```python
print(blankline_tokenize(s))
```

```python
from nltk.stem import PorterStemmer
from nltk.stem.lancaster import LancasterStemmer
from nltk.stem.snowball import SnowballStemmer
pst = PorterStemmer()
lst = LancasterStemmer()
print(lst.stem("eating"))
```

```python
print(pst.stem("shopping"))
```

```python
from nltk.stem import WordNetLemmatizer
wlem = WordNetLemmatizer()
print(wlem.lemmatize("ate"))
```

```python
from nltk.corpus import stopwords
stoplist = stopwords.words('english')
text = "This is just a test"
cleanwordlist = [word for word in text.split() if word not in stoplist]
print(cleanwordlist)
```

```python
from nltk.metrics import edit_distance
print(edit_distance("rain", "shine"))
```
