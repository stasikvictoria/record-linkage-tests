# Apprentisage de représentaion (Word Embeddings)

```python
from gensim.models import Word2Vec
from gensim.models import KeyedVectors
from sklearn.metrics.pairwise import cosine_similarity
```

```python
%pip install scikit-learn
```

```python
# Exemple de données
product_names = ["crème hydratante", "crème hydratante légère"]
```

```python
# Tokenisation des noms de produits
tokenized_names = [name.split() for name in product_names]
tokenized_names
```

```python
# Entraînement du modèle Word2Vec
# vector_size -> It defines the dimensionality of the word vectors. In other words, each word in the vocabulary will be represented as a vector of this size.
# window -> It represents the maximum distance between the current and predicted word within a sentence. A larger window allows the model to capture more context, but it may also lead to more computational cost.
# min_count -> Ignores all words with a total frequency lower than this. Words with very low frequency are often not meaningful for training reliable word vectors.
# workers -> The number of CPU cores to use when training the model. Increasing the number of workers can speed up the training process.
model = Word2Vec(sentences=tokenized_names, vector_size=100, window=5, min_count=1, workers=4)
```

```python
# Sauvegarde du modèle
model.wv.save_word2vec_format("product_embeddings.bin", binary=True)
```

```python
# Chargement du modèle sauvegardé
word_vectors = KeyedVectors.load_word2vec_format("product_embeddings.bin", binary=True)
word_vectors
```

```python
# Représentation vectorielle des noms de produits
vector1 = word_vectors["crème"] + word_vectors["hydratante"]
vector2 = word_vectors["crème"] + word_vectors["hydratante"] + word_vectors["légère"]
print(vector1)
```

```python
# Mesure de similarité cosinus
similarity = cosine_similarity([vector1], [vector2])
```

```python
#  Plus la similarité cosinus est proche de 1, plus les noms de produits sont similaires.
print("Similarité cosinus entre les noms de produits :", similarity[0, 0])
```

# Clustering (K-Means)

```python
from sklearn.cluster import KMeans
import numpy as np
```

```python
# Exemple de représentation vectorielle des noms de produits
product_vectors = np.array([[1.2, 0.5], [0.8, 1.0], [1.5, 0.7], [0.6, 0.9]])
```

```python
# Définissez le nombre de clusters (k)
num_clusters = 2
```

```python
# Appliquez K-means
kmeans = KMeans(n_clusters=num_clusters, random_state=42)
cluster_assignments = kmeans.fit_predict(product_vectors)
```

```python
print("Assignations des clusters :", cluster_assignments)
```

```python
# Example avec données fictives de produits
product_names = ["crème hydratante", "parfum jadore", "crème hydratante légère", "jadore dior"]
tokenized_names = [name.split() for name in product_names]
model = Word2Vec(sentences=tokenized_names, vector_size=100, window=5, min_count=1, workers=4)
model.wv.save_word2vec_format("product_embeddings.bin", binary=True)
word_vectors = KeyedVectors.load_word2vec_format("product_embeddings.bin", binary=True)
```

```python
from transformers import AutoTokenizer, AutoModel
import torch
from scipy.spatial.distance import cosine
```

```python
# Charger le modèle BERT pré-entraîné
model_name = "bert-base-uncased"  # Vous pouvez choisir un modèle BERT en fonction de votre langage
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)
```

```python
# Convertir les noms de produits en embeddings BERT
def get_bert_embedding(text):
    tokens = tokenizer(text, return_tensors="pt")
    with torch.no_grad():
        outputs = model(**tokens)
    embeddings = torch.mean(outputs.last_hidden_state, dim=1).squeeze(0)
    return embeddings.numpy()
```

```python
# Calculer la similarité cosinus entre les embeddings BERT
def cosine_similarity(embedding1, embedding2):
    return 1 - cosine(embedding1, embedding2)
```

```python
# Liste de noms de produits
product_names = ["crème hydratante", "parfum", "crème hydratante légère", "eau de parfum"]
```

```python
# Convertir les noms de produits en embeddings BERT
embeddings = [get_bert_embedding(name) for name in product_names]
```

```python
# Calculer la similarité entre les paires de produits
for i in range(len(product_names)):
    for j in range(i + 1, len(product_names)):
        similarity_score = cosine_similarity(embeddings[i], embeddings[j])
        print(f"Similarité entre '{product_names[i]}' et '{product_names[j]}': {similarity_score:.4f}")
```

# Réseau de neuronnes siamois

```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Embedding, Flatten, concatenate
from tensorflow.keras.optimizers import Adam
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
import numpy as np

# Liste de noms de produits
product_names = ["crème hydratante", "parfum", "crème hydratante légère", "eau de parfum"]

# Créer des paires de noms de produits et des labels de similarité
pairs = [(product_names[i], product_names[j], int(i == j)) for i in range(len(product_names)) for j in range(len(product_names))]

# Encoder les noms de produits
encoder = LabelEncoder()
encoded_product_names = encoder.fit_transform(product_names)

# Préparer les données pour l'entraînement
X_left = np.array([encoded_product_names[product_names.index(pair[0])] for pair in pairs])
X_right = np.array([encoded_product_names[product_names.index(pair[1])] for pair in pairs])
y = np.array([pair[2] for pair in pairs])

# Diviser les données en ensembles d'entraînement et de test
X_left_train, X_left_test, X_right_train, X_right_test, y_train, y_test = train_test_split(X_left, X_right, y, test_size=0.2, random_state=42)

# Créer la couche d'embedding partagée
embedding_dim = 50
embedding_layer = Embedding(input_dim=len(encoder.classes_), output_dim=embedding_dim, name="embedding")

# Branche gauche du réseau
left_input = Input(shape=(1,))
left_embedding = embedding_layer(left_input)
left_flatten = Flatten()(left_embedding)
left_dense = Dense(64, activation='relu')(left_flatten)

# Branche droite du réseau
right_input = Input(shape=(1,))
right_embedding = embedding_layer(right_input)
right_flatten = Flatten()(right_embedding)
right_dense = Dense(64, activation='relu')(right_flatten)

# Concaténer les deux branches
merged_layer = concatenate([left_dense, right_dense])

# Ajouter des couches supplémentaires si nécessaire
# ...

# Couche de sortie
output_layer = Dense(1, activation='sigmoid')(merged_layer)

# Créer le modèle
siamese_model = Model(inputs=[left_input, right_input], outputs=output_layer)

# Compiler le modèle
siamese_model.compile(loss='binary_crossentropy', optimizer=Adam(lr=0.001), metrics=['accuracy'])

# Entraîner le modèle
siamese_model.fit([X_left_train, X_right_train], y_train, epochs=10, batch_size=16, validation_data=([X_left_test, X_right_test], y_test))
```

```python
# Évaluation du modèle sur les données de test
evaluation = siamese_model.evaluate([X_left_test, X_right_test], y_test, verbose=2)

# Affichage des métriques
print("Loss:", evaluation[0])
print("Accuracy:", evaluation[1])
```

```python
# Exemple de prédiction pour une nouvelle paire de noms de produits
new_pair = (encoded_product_names[encoder.transform(["crème hydratante"])[0]],
            encoded_product_names[encoder.transform(["crème hydratante légère"])[0]])

# Faire la prédiction
prediction = siamese_model.predict([np.array([new_pair[0]]), np.array([new_pair[1]])])

# Afficher la prédiction
print("Prediction:", prediction)
```
