```python
from urllib.request import urlopen
html = urlopen("http://python.org/").read()
print(len(html))
```

```python
tokens = [tok for tok in html.split()]
print ("Total no of tokens : " + str(len(tokens)))
```

```python
print(tokens[0:100])
```

```python
import re
html_string = html.decode('utf-8')
tokens = re.split('\W+', html_string)
print(len(tokens))
```

```python
print(tokens[0:100])
```

```python
from bs4 import BeautifulSoup
import nltk
```

```python
soup = BeautifulSoup(html, 'html.parser')
clean = soup.get_text()
tokens = [tok for tok in clean.split()]
print(tokens[:100])
```

```python
import operator
freq_dis={}
for tok in tokens:
    if tok in freq_dis:
        freq_dis[tok] += 1
```
